# Results tables

The paper's tables, in order: the precomputation, with and without precomputation,
sampler and $\epsilon$, and the models compared. Each averages over its seeds. The log
parsing and LaTeX building live in `results_tables.py`.

```python
%load_ext autoreload
%autoreload 2

from results_tables import (eps_column_latex, load, model_comparison_latex,
                            precomp_comparison_latex, precomp_disagreement,
                            precomputation_latex, read_baseline, show)
```

```python
# Descending, as the epsilon table's rows read.
by_eps = {eps: load(epsilon=eps) for eps in (0.2, 0.1, 0.05, 0.01)}
empty = read_baseline("empty-tbox")
reasoner = read_baseline("reasoner-precomp")
```

## 1. Precomputation

The LLM's precomputation against the reasoner's, which answers every class pair
correctly (`logs/baselines/reasoner-precomp.log`), with the empty T-box as the floor.

```python
precomputation_tex = precomputation_latex(by_eps[0.2], models=["mistral"], empty=empty, reasoner=reasoner)
show(precomputation_tex)
#print(precomputation_tex)
```

```python
wrong, missed = precomp_disagreement(by_eps[0.2], "mistral", reasoner)
print(f"{len(wrong)} added but not entailed, {len(missed)} entailed but not added")
for sub, sup in wrong:
    print(f"  wrong:  {sub} SubClassOf {sup}")
for sub, sup in missed:
    print(f"  missed: {sub} SubClassOf {sup}")
```

## 2. With and without precomputation

```python
precomp_comparison_tex = precomp_comparison_latex(by_eps[0.2], model="mistral", sampler="unweighted")
show(precomp_comparison_tex)
#print(precomp_comparison_tex)
```

## 3. Sampler and $\epsilon$

$\epsilon$ sets the PAC sample budget: 10,661 samples at 0.2, 21,322 at 0.1 and
213,449 at 0.01. Every base set has its ten seeds at 0.2 and at 0.01 under both
samplers -- the C2 reruns, lost the first time to the shared query cache, landed on
2026-09-17. The 0.1 arm was run unweighted only.

```python
eps_column_tex = eps_column_latex(by_eps, model="mistral")
show(eps_column_tex)
#print(eps_column_tex)
```

## 4. Models compared

One setting, every model run at it, against the empty T-box.

```python
models_tex = model_comparison_latex(by_eps[0.1], baseset="C2", sampler="unweighted", empty=empty)
show(models_tex)
#print(models_tex)
```
